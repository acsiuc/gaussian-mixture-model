```python
import numpy as np

rng = np.random.default_rng(seed=0)
group1 = rng.normal(loc=165, scale=6, size=500)

print(group1[:10])
```

```python
print(group1.mean())
```

```python
print(group1.std())
```

```python
group2 = rng.normal(loc = 180, scale = 6, size = 500)
print(group2.mean())
```

```python
one_labels = np.ones(500)
zero_labels = np.zeros(500)
labels = np.concatenate([zero_labels, one_labels])
heights = np.concatenate([group1, group2])
```

```python
shuffle_order = rng.permutation(1000)

labels_shuffled = labels[shuffle_order]
heights_shuffled = heights[shuffle_order]
```

```python
print(heights_shuffled[:10])
print(labels_shuffled[:10])
```
