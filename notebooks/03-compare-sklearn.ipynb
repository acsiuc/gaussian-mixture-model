```python
import numpy as np
from sklearn.mixture import GaussianMixture

from gaussian_mixture_model.data import make_two_groups
from gaussian_mixture_model.gmm import GMM

heights, labels = make_two_groups(165, 180, 6, 500, seed=0)

# our GMM
model = GMM(k_curves=2, tolerance=0.001, seed=0, max_rounds=1000)
model.fit(heights)

# scikit-learn's GMM
sk = GaussianMixture(n_components=2, random_state=0)
sk.fit(heights.reshape(-1, 1))

print("ours    means:", model.means_.round(2))
print("sklearn means:", sk.means_.ravel().round(2))
print("ours    stds: ", model.stds_.round(2))
print("sklearn stds: ", np.sqrt(sk.covariances_.ravel()).round(2))
print("ours    weights:", model.weights_.round(2))
print("sklearn weights:", sk.weights_.round(2))
```
